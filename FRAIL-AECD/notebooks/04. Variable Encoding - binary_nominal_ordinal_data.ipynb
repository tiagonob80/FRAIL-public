```python
def encode_binary_columns(df):
    # Genero
    df["Genero"] = df["Genero"].replace({"F": 0, "M": 1}).fillna(-1).astype(int)

    # Quedas_ultimos_12meses
    df["Quedas_ultimos_12meses"] = df["Quedas_ultimos_12meses"].replace({"não": 0, "sim": 1}).fillna(-1).astype(int)

    # Quedas_12meses_Lesao_Sim_Nao
    df["Quedas_12meses_Lesao_Sim_Nao"] = df["Quedas_12meses_Lesao_Sim_Nao"].replace({"não": 0, "sim": 1}).fillna(-1).astype(int)

    # Quedas_12meses_Assistencia_hospitalar_Sim_Não
    df["Quedas_12meses_Assistencia_hospitalar_Sim_Não"] = df["Quedas_12meses_Assistencia_hospitalar_Sim_Não"].replace({"não": 0, "sim": 1}).fillna(-1).astype(int)

    # Quedas_12meses_Hospitalizado_Sim_Não
    df["Quedas_12meses_Hospitalizado_Sim_Não"] = df["Quedas_12meses_Hospitalizado_Sim_Não"].replace({"não": 0, "sim": 1}).fillna(-1).astype(int)

    # Quedas_12meses_Medo_cair_Sim_Não
    df["Quedas_12meses_Medo_cair_Sim_Não"] = df["Quedas_12meses_Medo_cair_Sim_Não"].replace({"não": 0, "sim": 1}).fillna(-1).astype(int)

    # ATIVIDADE_FISICA_Geral_Sim_Não
    df["ATIVIDADE_FISICA_Geral_Sim_Não"] = df["ATIVIDADE_FISICA_Geral_Sim_Não"].replace({"não": 0, "sim": 1}).fillna(-1).astype(int)

    # HANDGRIP_Mao_DOMINANTE
    df["HANDGRIP_Mao_DOMINANTE"] = df["HANDGRIP_Mao_DOMINANTE"].replace({"esquerda": 0, "direita": 1, "ambidestro": 1}).fillna(-1).astype(int)

    # TIMED_UP_AND_GO_Simples_Apoio_mãos
    df["TIMED_UP_AND_GO_Simples_Apoio_mãos"] = df["TIMED_UP_AND_GO_Simples_Apoio_mãos"].replace({"não": 0, "sim": 1}).fillna(-1).astype(int)

    # TIMED_UP_AND_GO_Simples_Auxiliar_marcha
    df["TIMED_UP_AND_GO_Simples_Auxiliar_marcha"] = df["TIMED_UP_AND_GO_Simples_Auxiliar_marcha"].replace({"não": 0, "sim": 1}).fillna(-1).astype(int)

    # TIMED_UP_AND_GO_Dupla_tarefa_Apoio_maos
    df["TIMED_UP_AND_GO_Dupla_tarefa_Apoio_maos"] = df["TIMED_UP_AND_GO_Dupla_tarefa_Apoio_maos"].replace({"não": 0, "sim": 1}).fillna(-1).astype(int)

    # TIMED_UP_AND_GO_Dupla_tarefa_Auxiliar_marcha
    df["TIMED_UP_AND_GO_Dupla_tarefa_Auxiliar_marcha"] = df["TIMED_UP_AND_GO_Dupla_tarefa_Auxiliar_marcha"].replace({"não": 0, "sim": 1}).fillna(-1).astype(int)

    # SIT_TO_STAND_Apoio_maos
    df["SIT_TO_STAND_Apoio_maos"] = df["SIT_TO_STAND_Apoio_maos"].replace({"não": 0, "sim": 1}).fillna(-1).astype(int)

    # SIT_TO_STAND_Auxiliar_marcha
    df["SIT_TO_STAND_Auxiliar_marcha"] = df["SIT_TO_STAND_Auxiliar_marcha"].replace({"não": 0, "sim": 1}).fillna(-1).astype(int)

    # @6_MIN_ANDAR_Auxiliar_marcha
    df["@6_MIN_ANDAR_Auxiliar_marcha"] = df["@6_MIN_ANDAR_Auxiliar_marcha"].replace({"não": 0, "sim": 1}).fillna(-1).astype(int)

    # @6_MIN_ANDAR_Parou
    df["@6_MIN_ANDAR_Parou"] = df["@6_MIN_ANDAR_Parou"].replace({"não": 0, "sim": 1}).fillna(-1).astype(int)


    return df
```

```python
def encode_categorical_columns(df):
    # ID_Aluno
    df["ID_Aluno"] = df["ID_Aluno"].fillna(-1).astype(int)

    # Momento
    df["Momento"] = df["Momento"].fillna(-1).astype(int)

    # Local_de_Avaliacao
    df["Local_de_Avaliacao"] = df["Local_de_Avaliacao"].replace({"Famalicão": 0, "Joane": 1, "Ribeirão": 2, "São Mateus": 3}).fillna(-1).astype(int)

    # Modalidade
    df["Modalidade"] = df["Modalidade"].replace({"Terra": 0, "Água": 1, "Água + Terra": 2}).fillna(-1).astype(int)

    # @6_MIN_ANDAR_Percurso
    df["@6_MIN_ANDAR_Percurso"] = df["@6_MIN_ANDAR_Percurso"].replace({"A": 0, "B": 1, "C": 2, "D": 3, "E": 4, "F": 5}).fillna(-1).astype(int)

    return df
```

## Encode all data function

```python
data.head()
```

```python
def encode_all_data(df):
    df = encode_binary_columns(df)
    df = encode_categorical_ordinal_columns(df)
    df = encode_categorical_columns(df)

    df = df.fillna(-1)
    df = df.replace({-1: np.nan})


    return df

data = encode_all_data(data)
data.head()
```
