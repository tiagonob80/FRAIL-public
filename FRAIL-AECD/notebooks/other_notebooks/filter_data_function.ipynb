## Function

```python
import pandas as pd

def filter_data(data: pd.DataFrame, min_instances: int, sequential: bool, aligned: list[int] = None) -> pd.DataFrame:
    """
    Filters users based on number of test instances, sequential constraint, and optional alignment.

    Args:
        data (pd.DataFrame): Dataset with 'ID_Aluno' and 'Momento' columns.
        min_instances (int): Required number of test instances (1–4).
        sequential (bool): If True, instances must form a sequential group (e.g., [1,2,3]).
        aligned (list[int], optional): If provided, user must have exactly these moments. Must match min_instances length.

    Returns:
        pd.DataFrame: Filtered data meeting the criteria.

    Behavior:
        - If aligned is given: selects users matching those moments only.
        - If sequential is True: finds the last available sequential group of min_instances moments.
        - Otherwise: returns last min_instances moments for users with at least that many.
    """
    if aligned:
        if len(aligned) != min_instances:
            raise ValueError("Length of aligned must equal min_instances.")
        if sorted(aligned) != list(range(min(aligned), max(aligned) + 1)) and sequential:
            raise ValueError("Aligned moments are not sequential, so sequential must be False.")

    filtered_rows = []

    for user_id, group in data.groupby("ID_Aluno"):
        momentos = sorted(group["Momento"].unique())

        if len(momentos) < min_instances:
            continue

        if aligned:
            if all(m in momentos for m in aligned):
                filtered_rows.append(group[group["Momento"].isin(aligned)])
        elif sequential:
            # Search for the last sequential block of min_instances
            for i in reversed(range(len(momentos) - min_instances + 1)):
                window = momentos[i:i + min_instances]
                if window == list(range(window[0], window[0] + min_instances)):
                    subset = group[group["Momento"].isin(window)]
                    filtered_rows.append(subset)
                    break
        else:
            # No sequential constraint, just last N
            last_moments = sorted(momentos)[-min_instances:]
            subset = group[group["Momento"].isin(last_moments)]
            filtered_rows.append(subset)

    return pd.concat(filtered_rows, ignore_index=True)
```
